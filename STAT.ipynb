3.A dataset has a mean of 50 and a standard deviation of 5. What percentage of values lie between 45 and 55 if the data follows a normal distribution?
Given:
- Mean (\(\mu\)) = 50
- Standard deviation (\(\sigma\)) = 5
- Lower value = 45
- Upper value = 55
Step 1: Calculate the Z-scores.
Formula:
\[
Z=\frac{X-\mu}{\sigma}
\]

For 45:
\[
Z=\frac{45-50}{5}=-1
\]

For 55:
\[
Z=\frac{55-50}{5}=1
\]

Step 2: In a normal distribution, approximately 68.27% of values lie within one standard deviation of the mean.
Excel formula:
=NORM.DIST(55,50,5,TRUE)-NORM.DIST(45,50,5,TRUE)


Answer: 68.27%



4.What is the concept of standardization (z-score), and why is it important in data analysis? Explain the
formula and how standardization transforms a dataset.
Standardization is the process of converting data values into Z-scores so that we can understand how far each value lies from the mean in terms of standard deviations.
Formula:
\[
Z=\frac{X-\mu}{\sigma}
\]

Where:
- \(X\) = Individual data value
- \(\mu\) = Mean of the dataset
- \(\sigma\) = Standard deviation
Example:Formula:
\[
Z=\frac{X-\mu}{\sigma}
\]

Where:
- \(X\) = Individual data value
- \(\mu\) = Mean of the dataset
- \(\sigma\) = Standard deviation
Example:Suppose the marks of a student are 80, the mean is 70, and the standard deviation is 5.
\[
Z=\frac{80-70}{5}=2
\]

The Z-score is 2, meaning the student's marks are 2 standard deviations above the mean.
Importance of standardization:
1. It helps compare values from different datasets.
2. It puts data on a common scale.
3. It helps identify outliers.
4. It is useful in machine learning and statistical analysis.
Conclusion: Standardization transforms data into Z-scores with a mean of 0 and a standard deviation of 1.


5.What is Kurtosis and their type?
Kurtosis is a statistical measure that describes the heaviness of a distribution's tails and its tendency to produce extreme values.
There are three main types of kurtosis:
1. Leptokurtic
Has heavier tails than a normal distribution and often a sharper peak. It has positive excess kurtosis.
2. Mesokurtic
Has the same kurtosis as a normal distribution. Its excess kurtosis is zero.
3. Platykurtic
Has lighter tails than a normal distribution and typically a flatter peak. It has negative excess kurtosis.
Conclusion: Kurtosis helps us understand the tail behavior of a dataset and its potential for extreme values.

6.Explain why the uniform distribution is a good model for the outcome of rolling a fair die.
A uniform distribution is appropriate for a fair die because all six outcomes have an equal probability of occurring.
The possible outcomes are:
\[
1,2,3,4,5,6
\]

The probability of each outcome is:
\[
P(X=x)=\frac{1}{6}
\]

Therefore:
\[
P(1)=P(2)=\cdots=P(6)=\frac{1}{6}
\]

Each outcome has approximately a 16.67% chance of occurring on a single roll.
Conclusion: The uniform distribution models a fair die because every face has an equal chance of appearing.

8.How does log transformation help in stabilizing variance and making data more normally distributed?
Log transformation is a technique in which we replace each data value with its logarithm.
Formula:
\[
Y=\log(X)
\]

In Excel, use:=LN(A1)
This calculates the natural logarithm of the value in cell A1. The original value must be positive.
Benefits of log transformation:
1. Reduces skewness: It can make positively skewed data more symmetric.
2. Stabilizes variance: It can reduce the effect of large values when variability increases with the mean.
3. Reduces the impact of extreme values: Large values become less dominant.
4. Improves analysis: It can help meet the assumptions of some statistical methods.
Example: Income data often contains a few very high values. Applying a logarithmic transformation reduces the differences between these large values and smaller values.
Conclusion: Log transformation can make some datasets more symmetric and stabilize variance, but it does not guarantee a normal distribution.
